```python
omama_loader = O.OmamaLoader()
data = O.Data(data_loader=omama_loader, load_cache=True)
```

```python
img1 = O.DataHelper.get(image='DXm.2.25.106924144890407873251702717161766133956')
img2 = O.DataHelper.get(image='DXm.2.25.109865729760992638360694443409251815567')
img3 = O.DataHelper.get(image='DXm.2.25.173803440485307451614987519907264671779')
img4 = O.DataHelper.get(image='DXm.2.25.213465259260888069289877692759291753499')
```

```python
import json

with open('/raid/mpsych/OMAMA/2025/PLAYGROUND/out/2.25.151222280059972160195646752408514211111/results_full.json') as f:
    results = json.load(f)
```

```python
for img in [img1, img2, img3, img4]:
    sop_uid = img.SOPInstanceUID
    if sop_uid in results['results']['dicom_results']:
        coords = results['results']['dicom_results'][sop_uid]['none'][0]['coords']
        print(f"Prediction for {sop_uid} → Score: {results['results']['dicom_results'][sop_uid]['none'][0]['score']}")
        O.DataHelper.view(image=img, roi_coords=coords)
    else:
        print(f"No predictions for image {sop_uid}")
```
