```python
import pydicom, matplotlib.pyplot as plt

for i, p in enumerate(paths, 1):
    ds = pydicom.dcmread(str(p), force=True)
    plt.subplot(2, 3, i)
    plt.imshow(ds.pixel_array, cmap="gray")
    plt.title(f"{ds.ImageLaterality} | {ds.ViewPosition}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
