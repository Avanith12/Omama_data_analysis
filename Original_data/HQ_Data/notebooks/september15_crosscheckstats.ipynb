```python
from pathlib import Path
from collections import Counter
import json, pickle

ROOT = Path("/home/avanith/Documents/projects/omama new data")
META = ROOT / "256x256_studylevel/metadata"
IMGS = ROOT / "256x256_studylevel/images"
PKL = ROOT / "2d_studylevel_all_dicom_headers.pkl"
```

```python
lat = [json.loads(p.read_text())["ImageLaterality"] for p in META.rglob("*.json")]
print(set(lat))
print(Counter(lat))
```

```python
view = [json.loads(p.read_text())["View"] for p in META.rglob("*.json")]
print(set(view))
print(Counter(view))
```

```python
headers = pickle.load(open(PKL, "rb"))
print(len(headers))
```

```python
lat = [ds.ImageLaterality for ds in headers.values()]
print(set(lat))
print(Counter(lat))
```

```python
view = [ds.ViewPosition for ds in headers.values()]
print(set(view))
print(Counter(view))
```
