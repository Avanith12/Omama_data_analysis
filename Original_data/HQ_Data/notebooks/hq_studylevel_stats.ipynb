```python
import json
from pathlib import Path

ROOT = Path("/hpcstor6/scratch01/a/a.kanamarlapudi001/HQ_StudyLevel_Data/original_studylevel/metadata")
```

```python
from collections import Counter

stats = {
    "Image Count": 0,
    "Patient Count": 0,
    "Study Count": 0,
    "Cancer Count": 0,
    "Deepsight Valid Score": 0,  # image-level: cancer images with score >= 0.5
    "CC Count": 0,
    "MLO Count": 0,
    "L Count": 0,
    "R Count": 0,
}
view_names = Counter()
patients = set()
studies = set()

for study_dir in ROOT.iterdir():
    if not study_dir.is_dir():
        continue
    studies.add(study_dir.name)
    for jp in study_dir.glob("*.json"):
        d = json.loads(jp.read_text())
        stats["Image Count"] += 1
        patients.add(d["PatientID"])

        view = d.get("View", "")
        view_names[view] += 1
        if view == "medio-lateral oblique":
            stats["MLO Count"] += 1
        elif view in ("cranio-caudal", "cranio-caudal exaggerated laterally"):
            stats["CC Count"] += 1

        if d.get("ImageLaterality") == "L":
            stats["L Count"] += 1
        if d.get("ImageLaterality") == "R":
            stats["R Count"] += 1

        if d.get("label") in ("IndexCancer", "PreIndexCancer"):
            stats["Cancer Count"] += 1
            if float(d.get("score", 0)) >= 0.5:
                stats["Deepsight Valid Score"] += 1

stats["Patient Count"] = len(patients)
stats["Study Count"] = len(studies)
print(view_names)
stats
```
