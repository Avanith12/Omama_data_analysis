```python
from pathlib import Path

import pydicom
import matplotlib.pyplot as plt
import numpy as np

STUDY = "/raid/data01/deephealth/dh_dcm_ast/2.25.78129393927635456805162825461213351679"

l_cc = "/raid/data01/deephealth/dh_dcm_ast/2.25.78129393927635456805162825461213351679/DXm.2.25.299041631368278452157869334748449288246"
r_cc = "/raid/data01/deephealth/dh_dcm_ast/2.25.78129393927635456805162825461213351679/DXm.2.25.22159323450079300213014046205604036099"
l_mlo = "/raid/data01/deephealth/dh_dcm_ast/2.25.78129393927635456805162825461213351679/DXm.2.25.51399802538781866645998426577062213753"
r_mlo = "/raid/data01/deephealth/dh_dcm_ast/2.25.78129393927635456805162825461213351679/DXm.2.25.187603759254809804197680356071706456026"

print("folder:", STUDY)
print("files:")
for f in sorted(Path(STUDY).glob("DXm.*")):
    print(" ", f.name)

print("\n4 views:")
print("l-cc ", l_cc)
print("r-cc ", r_cc)
print("l-mlo", l_mlo)
print("r-mlo", r_mlo)
```

```python
# l-cc
ds = pydicom.dcmread(l_cc, force=True)
plt.figure(figsize=(6, 6))
plt.imshow(ds.pixel_array, cmap="gray")
plt.title("l-cc")
plt.axis("off")
plt.show()
```
