#  CombinedvsSingle4_dh_dhnew_Grep Cancer

##  Study Selection Description

The following study IDs are extracted from the `dh_dh0new_labels.csv` file located in `/raid/data01/deephealth/labels/`.  
Only studies labeled with `IndexCancer` were selected for evaluation.  
Each selected study has **at least 4 DICOM images**, and is located in the `dh_dh0new` dataset.

These studies are used to compare **DeepSight detection scores** between:
- A combined run (using 4 images together)
- Individual single-image runs (img0 to img3)

The total and laterality-specific scores (L and R) are printed for each.


```python
# Import Required Libraries
import os
import json

# Base path to DeepSight output
BASE_DIR = "/raid/mpsych/OMAMA/2025/PLAYGROUND/DeepSight_CancerOnly"

# List study IDs (strip _combined/_img*)
all_entries = os.listdir(BASE_DIR)
study_ids = sorted(list(set(e.split("_")[0] for e in all_entries)))

# We'll only look at the first 10
study_ids = study_ids[:10]
```

# Function to Load Combined Scores

This helper function reads a `results_full.json` file and extracts scores from the `study_results` section.
Used for both combined and single image runs.

```python
def load_scores(json_path):
    try:
        with open(json_path, "r") as f:
            data = json.load(f)
        study = data.get("results", {}).get("study_results", {})
        total = round(study.get("total", {}).get("score", 0), 4)
        L = round(study.get("L", {}).get("score", 0), 4)
        R = round(study.get("R", {}).get("score", 0), 4)
        return total, L, R
    except:
        return None, None, None
```

# Alternative Score Extractor (from raw DICOM results)

This function parses the `results_raw.dicom_results` field directly. 
It handles edge cases where `study_results` may be missing, ensuring partial recovery of total, L, and R scores.


```python
def extract_scores(json_path):
    if not os.path.exists(json_path):
        return None, None, None
    try:
        with open(json_path) as f:
            data = json.load(f)
        if not data or "results_raw" not in data:
            return None, None, None
        dicoms = data["results_raw"].get("dicom_results", {})
        total, L, R = 0.0, 0.0, 0.0
        for dicom in dicoms.values():
            for obj in dicom.get("none", []):
                score = obj.get("score", 0.0)
                lat = obj.get("laterality")
                if lat == "L": L += score
                elif lat == "R": R += score
                else: total += score
        return round(total, 4), round(L, 4) if L else None, round(R, 4) if R else None
    except Exception as e:
        print(f" Error parsing {json_path}: {e}")
        return None, None, None
```

# View Scores for First Study in Table Format

This cell:
- Loads the first study ID.
- Extracts and prints the Combined and 4 Single image scores.
- Displays results in a clean table format.


```python
sid = study_ids[0]  

print(f"\n===== Study: {sid} =====")
rows = []

# Combined
comb_json = os.path.join(BASE_DIR, f"{sid}_combined", sid, "results_full.json")
ct, cl, cr = load_scores(comb_json)
rows.append(("Combined", ct, cl, cr))

# Singles
for i in range(4):
    single_json = os.path.join(BASE_DIR, f"{sid}_img{i}", sid, "results_full.json")
    st, sl, sr = load_scores(single_json)
    rows.append((f"single/0{i+1}", st, sl, sr))

# Print Table
print(f"{'Run':<10} {'Total':<10} {'L':<10} {'R':<10}")
for r in rows:
    print(f"{r[0]:<10} {r[1]:<10} {r[2]:<10} {r[3]:<10}")
```
