```python
#This the for 1 patient 
# === IMPORTS ===
import os
import subprocess
import tempfile

# === CONFIGURATION ===
base_dir = "/raid/data01/deephealth/dh_dh0new/"  # Now using dh_dh0new as input directly!
output_base = "/raid/mpsych/OMAMA/2025/PLAYGROUND/DeepSight_Outputs/"
deepsight_script = "/home2/deephealth/scripts/deepsight2.sh"

# === SETUP ===
os.makedirs(output_base, exist_ok=True)

studies = sorted(os.listdir(base_dir))[:1]  # Only run on 1 patient now for test

# === MAIN LOOP ===
for study in studies:
    study_path = os.path.join(base_dir, study)
    if not os.path.isdir(study_path):
        continue

    dicoms = [f for f in os.listdir(study_path) if f.startswith("DXm")]
    dicom_paths = [os.path.join(study_path, d) for d in dicoms]

    if len(dicom_paths) < 4:
        print(f"Skipping {study} (not enough images)")
        continue

    # Run 4 images together
    with tempfile.NamedTemporaryFile(mode="w", delete=False, suffix=".txt") as tmpfile:
        for path in dicom_paths:
            tmpfile.write(path + "\n")
        caselist_path = tmpfile.name

    output_folder = os.path.join(output_base, f"{study}_4img")
    os.makedirs(output_folder, exist_ok=True)

    cmd = [
        deepsight_script,
        "-i", base_dir,
        "-o", output_folder,
        "-cl", caselist_path
    ]
    print(f"Running DeepSight on {study} (4 images)")
    subprocess.run(cmd)

    os.remove(caselist_path)

    # Run each image individually
    for idx, dicom_path in enumerate(dicom_paths):
        with tempfile.NamedTemporaryFile(mode="w", delete=False, suffix=".txt") as tmpfile_single:
            tmpfile_single.write(dicom_path + "\n")
            caselist_path_single = tmpfile_single.name

        output_folder_single = os.path.join(output_base, f"{study}_img{idx}")
        os.makedirs(output_folder_single, exist_ok=True)

        cmd_single = [
            deepsight_script,
            "-i", base_dir,
            "-o", output_folder_single,
            "-cl", caselist_path_single
        ]
        print(f"Running DeepSight on {study} (single image {idx+1})")
        subprocess.run(cmd_single)

        os.remove(caselist_path_single)

print("✅ TEST COMPLETE! DeepSight ran successfully on 1 patient.")
```

```python
import os
import json

# === CONFIGURATION ===
output_base = "/raid/mpsych/OMAMA/2025/PLAYGROUND/DeepSight_Outputs_25patients_REANALYSIS/"

# === Collect Predictions ===
predictions = []

# List all patient folders
all_folders = sorted(os.listdir(output_base))

for folder in all_folders:
    folder_path = os.path.join(output_base, folder)

    # Check if it has a subfolder (study folder)
    subfolders = os.listdir(folder_path)
    for subfolder in subfolders:
        subfolder_path = os.path.join(folder_path, subfolder)

        # Look for results_full.json inside the study folder
        json_file_path = os.path.join(subfolder_path, "results_full.json")
        if os.path.exists(json_file_path):
            with open(json_file_path, "r") as f:
                data = json.load(f)

            # Parse prediction score
            try:
                score = data["results_raw"]["dicom_results"]
                for sop_uid, sop_data in score.items():
                    if sop_data["none"]:
                        pred_score = sop_data["none"][0]["score"]
                        predictions.append({
                            "folder": folder,
                            "sop_uid": sop_uid,
                            "score": pred_score
                        })
            except Exception as e:
                print(f" Error reading {json_file_path}: {e}")

print("\n✅ Finished collecting predictions!")
print(f"Total predictions found: {len(predictions)}\n")

# === Print some predictions nicely ===
for pred in predictions[:20]:  # Show first 20 only
    print(f"Folder: {pred['folder']} | SOP UID: {pred['sop_uid']} | Score: {pred['score']}")

print("\n DONE!")
```

```python
# === Quick debug check: what SOPs exist in 4img and 1img modes ===

sop_4img = set()
sop_1img = set()

for pred in predictions:
    folder = pred['folder']
    sop_uid = pred['sop_uid']

    if folder.endswith("_4img"):
        sop_4img.add(sop_uid)
    elif "_img" in folder:
        sop_1img.add(sop_uid)

print(f"Total unique SOPs in 4-image mode: {len(sop_4img)}")
print(f"Total unique SOPs in 1-image mode: {len(sop_1img)}")

common_sops = sop_4img.intersection(sop_1img)
print(f"✅ SOPs common between 4img and 1img: {len(common_sops)}")
```
