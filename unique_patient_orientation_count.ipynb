```python
import collections
import omama as O

# Load the dataset
data = O.Data(O.OmamaLoader(config_num=2), load_cache=True)

# Dictionary to track orientations for each patient
patient_orientation_map = collections.defaultdict(set)

# Iterate through all images and store orientations per patient
for image_path in data._image_paths:
    ds = O.Data._dicom(dicom_path=image_path, dicom_header=True)  # Load DICOM metadata
    patient_id = ds.PatientID
    orientation = ds.ImageLaterality + "-" + ds.ViewPosition  # e.g., "L-CC", "R-MLO"
    
    if orientation in {"L-CC", "L-MLO", "R-CC", "R-MLO"}:  # Only keep relevant orientations
        patient_orientation_map[patient_id].add(orientation)

# Count patients with all 4 orientations
patients_with_all_4 = [pid for pid, orientations in patient_orientation_map.items() if len(orientations) == 4]

# Print results
print(f"Total unique patients in dataset: {len(patient_orientation_map)}")
print(f"Patients with all 4 orientations: {len(patients_with_all_4)}")

# If needed, save patient list for further analysis
with open("patients_with_all_4_orientations.txt", "w") as f:
    for pid in patients_with_all_4:
        f.write(f"{pid}\n")
```

```python
import collections
import omama as O

# Load the dataset
data = O.Data(O.OmamaLoader(config_num=2), load_cache=True)

# Store found orientation names
found_orientations = set()

# Iterate through all images and collect orientation names
for image_path in data._image_paths:
    ds = O.Data._dicom(dicom_path=image_path, dicom_header=True)  # Load DICOM metadata
    
    laterality = ds.get("ImageLaterality")  # Expected: "L" or "R"
    view_position = ds.get("ViewPosition")  # Expected: "CC" or "MLO" (or full name)

    if laterality and view_position:
        orientation = f"{laterality}-{view_position}"  # e.g., "L-CC", "L-Craniocaudal"
        found_orientations.add(orientation)

# Print unique orientation names found
print("Orientation names found in dataset:")
print(found_orientations)
```
