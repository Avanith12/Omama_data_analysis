```python
import os
import glob
import shutil
import random
from tqdm import tqdm

random.seed(42)

# Source directory with .npz files
NPZ_DIR = "/hpcstor6/scratch01/a/a.kanamarlapudi001/balanced_dataset_from_2dresized/npz"
DEST_DIR = os.getcwd()

# Define destination split and class folders
splits = ["train_images", "val_images", "test_images"]
classes = ["Cancer", "NonCancer"]

# Create nested folders: split/class/
for split in splits:
    for cls in classes:
        os.makedirs(os.path.join(DEST_DIR, split, cls), exist_ok=True)

# Load and shuffle Cancer and NonCancer files
cancer_npz = sorted(glob.glob(os.path.join(NPZ_DIR, "Cancer", "*.npz")))
noncancer_npz = sorted(glob.glob(os.path.join(NPZ_DIR, "NonCancer", "*.npz")))

random.shuffle(cancer_npz)
random.shuffle(noncancer_npz)

# Function to split list
def split_files(files, train_ratio=0.6, val_ratio=0.2):
    total = len(files)
    train_end = int(train_ratio * total)
    val_end = train_end + int(val_ratio * total)
    return files[:train_end], files[train_end:val_end], files[val_end:]

# Split both classes
c_train, c_val, c_test = split_files(cancer_npz)
n_train, n_val, n_test = split_files(noncancer_npz)

# Mapping for copying
split_map = {
    "train_images": {"Cancer": c_train, "NonCancer": n_train},
    "val_images": {"Cancer": c_val, "NonCancer": n_val},
    "test_images": {"Cancer": c_test, "NonCancer": n_test}
}

# Copy files to split/class folders
for split, class_dict in split_map.items():
    for cls, file_list in class_dict.items():
        dest_path = os.path.join(DEST_DIR, split, cls)
        for file in tqdm(file_list, desc=f"{split}/{cls}"):
            shutil.copy(file, os.path.join(dest_path, os.path.basename(file)))

# Final summary
print("\n✅ Dataset successfully split and saved with class subfolders.")
```

```python
import os
import shutil
from tqdm import tqdm

# Config paths
DATASET_DIR = "dataset"
# Just replace the METADATA_SOURCE line:
METADATA_SOURCE = "/hpcstor6/scratch01/a/a.kanamarlapudi001/balanced_dataset_from_2dresized/metadata"
OUTPUT_FOLDER_NAMES = {
    "train": "train_metadata",
    "val": "val_metadata",
    "test": "test_metadata"
}

# Create destination folders
for split in OUTPUT_FOLDER_NAMES.values():
    os.makedirs(split, exist_ok=True)

# For each split (train/val/test)
for split_key, out_folder in OUTPUT_FOLDER_NAMES.items():
    print(f"\n🔍 Processing {split_key.upper()} split")

    for cls in ["Cancer", "NonCancer"]:
        split_class_dir = os.path.join(DATASET_DIR, f"{split_key}_images", cls)
        npz_files = [f for f in os.listdir(split_class_dir) if f.endswith(".npz")]

        for npz_file in tqdm(npz_files, desc=f"{split_key}/{cls}"):
            base_id = os.path.splitext(npz_file)[0]
            json_src = os.path.join(METADATA_SOURCE, base_id + ".json")
            json_dst = os.path.join(out_folder, base_id + ".json")

            if os.path.exists(json_src):
                shutil.copy(json_src, json_dst)
            else:
                print(f"⚠️ Missing metadata: {base_id}.json")
```
