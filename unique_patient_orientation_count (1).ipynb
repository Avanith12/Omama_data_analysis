```python
#for Patients with all 4 orientations
import collections
import omama as O

# Load the dataset
data = O.Data(O.OmamaLoader(config_num=2), load_cache=True)

# Dictionary to track orientations for each patient
patient_orientation_map = collections.defaultdict(set)

# Iterate through all images and store orientations per patient
for image_path in data._image_paths:
    ds = O.Data._dicom(dicom_path=image_path, dicom_header=True)  # Load DICOM metadata
    patient_id = ds.PatientID
    orientation = ds.ImageLaterality + "-" + ds.ViewPosition  # e.g., "L-CC", "R-MLO"
    
    if orientation in {"L-CC", "L-MLO", "R-CC", "R-MLO"}:  # Only keep relevant orientations
        patient_orientation_map[patient_id].add(orientation)

# Count patients with all 4 orientations
patients_with_all_4 = [pid for pid, orientations in patient_orientation_map.items() if len(orientations) == 4]

# Print results
print(f"Total unique patients in dataset: {len(patient_orientation_map)}")
print(f"Patients with all 4 orientations: {len(patients_with_all_4)}")

# If needed, save patient list for further analysis
with open("patients_with_all_4_orientations.txt", "w") as f:
    for pid in patients_with_all_4:
        f.write(f"{pid}\n")
```

```python
#Patient Analysis: CC and MLO View Presence Check
import collections
import omama as O

# Load the dataset
data = O.Data(O.OmamaLoader(config_num=2), load_cache=True)

# Dictionary to track CC and MLO orientations for each patient
patient_orientation_map = collections.defaultdict(set)

# Iterate through all images and store orientations per patient
for image_path in data._image_paths:
    ds = O.Data._dicom(dicom_path=image_path, dicom_header=True)  # Load DICOM metadata
    
    # Extract metadata safely
    patient_id = ds.PatientID
    laterality = ds.get("ImageLaterality")  # "L" or "R"
    view_position = ds.get("ViewPosition")  # "CC" or "MLO"
    
    if laterality and view_position:
        orientation = f"{laterality}-{view_position}"  # Example: "L-CC", "R-MLO"
        patient_orientation_map[patient_id].add(orientation)

# Count patients with at least one CC and one MLO on the same side
patients_with_cc_mlo = []

for pid, orientations in patient_orientation_map.items():
    has_left_cc = "L-CC" in orientations
    has_left_mlo = "L-MLO" in orientations
    has_right_cc = "R-CC" in orientations
    has_right_mlo = "R-MLO" in orientations

    # Check if the patient has at least one CC and one MLO from the same side
    if (has_left_cc and has_left_mlo) or (has_right_cc and has_right_mlo):
        patients_with_cc_mlo.append(pid)

# Print results
print(f"Total unique patients in dataset: {len(patient_orientation_map)}")
print(f"Patients with at least one CC and one MLO from the same side: {len(patients_with_cc_mlo)}")

# Save results for reference
with open("patients_with_cc_and_mlo.txt", "w") as f:
    for pid in patients_with_cc_mlo:
        f.write(f"{pid}\n")

print("Patient list saved successfully!")
```
