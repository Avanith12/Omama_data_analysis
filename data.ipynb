```python
from pathlib import Path
import json, numpy as np, matplotlib.pyplot as plt

```

## 2D images

```python
imgs = sorted(Path("/hpcstor6/scratch01/a/a.kanamarlapudi001/KAGGLE_CHALLENGE/2d/images").glob("*.npz"))
meta_dir = Path("/hpcstor6/scratch01/a/a.kanamarlapudi001/KAGGLE_CHALLENGE/2d/metadata")

```

```python
def show_one(i=0):
    p = imgs[i]
    x = np.load(p)["data"]
    meta = json.load(open(meta_dir / f"{p.stem}.json"))
    plt.figure(figsize=(6, 6))
    plt.imshow(x, cmap="gray")
    plt.axis("off")
    plt.show()
    print(meta)
```

```python
show_one(0)
show_one(1)
show_one(2)
show_one(3)
show_one(4)
show_one(5)
show_one(6)
show_one(7)
show_one(8)
show_one(9)
```

## 3D Images

```python
vols = sorted(Path("/hpcstor6/scratch01/a/a.kanamarlapudi001/KAGGLE_CHALLENGE/3d/images").glob("*.npz"))
meta3d_dir = Path("/hpcstor6/scratch01/a/a.kanamarlapudi001/KAGGLE_CHALLENGE/3d/metadata")
```

```python
def show_one_3d(i=0):
    p = vols[i]
    vol = np.load(p)["data"]  # shape (Z, H, W)
    z = vol.shape[0] // 2
    meta = json.load(open(meta3d_dir / f"{p.stem}.json"))
    plt.figure(figsize=(6, 6))
    plt.imshow(vol[z], cmap="gray")
    plt.axis("off")
    plt.show()
    print(meta)
```

```python
show_one_3d(0)
show_one_3d(1)
show_one_3d(2)
```
